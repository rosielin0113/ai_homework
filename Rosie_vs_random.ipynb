## Define agents

The random agent is taken straight from `example.ipynb`. `RosieAgent` is imported from `Rosie_agent.py`.

```python
# from example.ipynb
def act_random(board, player):
    """A random Connect-4 agent function."""
    return np.random.choice(np.where(board[0] == 0)[0])

# my agent
rosie = RosieAgent("Rosie", total_budget_seconds=290)
```

## Single game — Rosie (red) vs Random (yellow)

Same `players` configuration as `example.ipynb`, just with `rosie.act` instead of the second random agent.

```python
rosie.reset()  # always reset before a new game

players = [
    {
        "algo": rosie.act,
        "name": "Rosie",
        "player": +1,
        "args": {}
    },
    {
        "algo": act_random,
        "name": "Random Agent",
        "player": -1,
        "args": {}
    }
]

result, final_board, all_boards = truly_dynamic_environment(
    players, size=(6, 7), visual=True
)

visualize(final_board)
```

## Replay (slow visualisation)

```python
replay(all_boards)
```

## Optional: 10-game match (5 as red, 5 as yellow)

This mirrors the tournament format. Random has a coin-flip element so all 10 games will be different. Rosie should win all of them.

```python
rosie_wins = random_wins = draws = 0

# 5 games — Rosie as red
for i in range(5):
    rosie.reset()  # reset budget every game
    players = [
        {"algo": rosie.act,  "name": "Rosie",  "player": +1, "args": {}},
        {"algo": act_random, "name": "Random", "player": -1, "args": {}},
    ]
    result, _, _ = truly_dynamic_environment(players, size=(6, 7), visual=False)
    if   result['winner'] == +1: rosie_wins  += 1
    elif result['winner'] == -1: random_wins += 1
    else:                        draws       += 1

# 5 games — Rosie as yellow
for i in range(5):
    rosie.reset()
    players = [
        {"algo": act_random, "name": "Random", "player": +1, "args": {}},
        {"algo": rosie.act,  "name": "Rosie",  "player": -1, "args": {}},
    ]
    result, _, _ = truly_dynamic_environment(players, size=(6, 7), visual=False)
    if   result['winner'] == -1: rosie_wins  += 1
    elif result['winner'] == +1: random_wins += 1
    else:                        draws       += 1

print()
print(f"=== 10-game match result ===")
print(f"Rosie  wins: {rosie_wins} / 10")
print(f"Random wins: {random_wins} / 10")
print(f"Draws:       {draws} / 10")
```
